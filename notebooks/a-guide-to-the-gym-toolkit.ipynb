# FrozenLake Environment

- S denotes the starting state
- F denotes the frozen state
- H denotes the hole state
- G denotes the goal state

**Objective:** Get to the goal state from S without visiting the H state.
**Reward Attribution:** +1 for G, 0 otherwise

```python
#Creating and Rendering the environment
import gymnasium as gym

env = gym.make("FrozenLake-v1", render_mode="ansi")
env.reset()
print(env.render())
```

```python
#State space
print(env.observation_space)
```

```python
#Action space
print(env.action_space)
```

Action Encoding

1. 0 -> left
2. 1 -> down
3. 2 -> right
4. 3 -> up

```python
#env.P[state][action]
#Move right from state 0(S)
print(env.unwrapped.P[0][2])
#Move down from state 3(F)
print(env.unwrapped.P[3][1])
```

## An Episode

```python
# Episode

import gymnasium as gym

env = gym.make("FrozenLake-v1", render_mode="ansi")
state = env.reset()

print('Time Step 0 :')
print(env.render())

num_timesteps = 20

for t in range(num_timesteps):
    random_action = env.action_space.sample()
    observation, reward, done, truncated,info = env.step(random_action)
    # If we reach terminal state
    print (f'Time Step {t+1}')
    print(env.render())
    if done:
        print(f'Reached Terminal State {observation} at time step {t + 1}')
        break
```

## Series of Episodes

```python
import gymnasium as gym

env = gym.make("FrozenLake-v1", render_mode="human")

num_episodes = 10
num_timesteps = 20

for i in range(num_episodes):
    state, info = env.reset()
    print(f'Episode {i}, Time Step 0:')
    env.render()

    for t in range(num_timesteps):
        random_action = env.action_space.sample()
        observation, reward, done, truncated, info = env.step(random_action)
        print(f'Time Step {t + 1}')
        env.render()

        if done or truncated:
            print(f'Reached Terminal State {observation} at time step {t + 1}')
            break

env.close()
```

## The Cart-Pole Environment

- State space comprises continuous values
- space values here include cart position, cart velocity, pole angle, pole velocity at the tip
- cart position ranges from -4.8 to 4.8
- the cart velocity ranges from -$\infty$ to +$\infty$ 
- the value of the pole angle ranges from -0.418 radians to 0.418
- the value of the pole velocity at the tip ranges from -$\infty$ to +$\infty$ 
- the state space is an array of values of the form **[cart_position, cart_velocity, pole angle, pole velocity at tip]**
- the action space is discrete.
- 0 -> push cart to the left
- 1 -> push cart to the right
- Agent receives a reward of +1 every time the pole stands up on the cart.


```python
import gymnasium as gym

env = gym.make("CartPole-v1", render_mode="human")
state, info = env.reset()
print(state)
env.render()
env.close()
```

```python
#State space information
print(env.observation_space),
print(env.observation_space.high)
print(env.observation_space.low)
```

```python
#Action space
print(env.action_space)
```

```python
import gymnasium as gym

env = gym.make("CartPole-v1", render_mode="human")

num_episodes = 100
num_timesteps = 50

for i in range(num_episodes):
    Return = 0
    state, info = env.reset()
    print(f'Episode {i}, Time Step 0:')
    env.render()

    for t in range(num_timesteps):
        random_action = env.action_space.sample()
        observation, reward, done, truncated, info = env.step(random_action)
        Return = Return + reward
        print(f'Time Step {t + 1}')
        env.render()

        if done or truncated:
            print(f'Reached Terminal State {observation} at time step {t + 1}')
            break
    if i%10==0:
        print(f'Episode: {i}, Return: {Return}')

env.close()
```
