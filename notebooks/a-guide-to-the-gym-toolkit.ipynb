# FrozenLake Environment

- S denotes the starting state
- F denotes the frozen state
- H denotes the hole state
- G denotes the goal state

**Objective:** Get to the goal state from S without visiting the H state.
**Reward Attribution:** +1 for G, 0 otherwise

## An Episode

```python
# Episode

import gymnasium as gym

env = gym.make("FrozenLake-v1", render_mode="ansi")
state = env.reset()

print('Time Step 0 :')
print(env.render())

num_timesteps = 20

for t in range(num_timesteps):
    random_action = env.action_space.sample()
    observation, reward, done, truncated,info = env.step(random_action)
    # If we reach terminal state
    print (f'Time Step {t+1}')
    print(env.render())
    if done:
        print(f'Reached Terminal State {observation} at time step {t + 1}')
        break
```

- 1 -> push cart to the right

## Value Iteration

```python
import numpy as np


def optimum_value(env, num_iterations=1000, threshold=1e-20, gamma=1.0):
    # Initialize value table
    value_table = np.zeros(env.observation_space.n)

    for i in range(num_iterations):
        # On every iteration, use the updated value table
        updated_value_table = np.copy(value_table)

        for s in range(env.observation_space.n):
            Q_values = [
                sum([prob * (r + gamma * updated_value_table[s_])
                     for prob, s_, r, _ in env.unwrapped.P[s][a]])
                for a in range(env.action_space.n)
            ]
            value_table[s] = max(Q_values)

        if np.sum(np.fabs(updated_value_table - value_table)) <= threshold:
            break

    return value_table


def extract_policy(value_table, env, gamma=1.0):
    policy = np.zeros(env.observation_space.n)

    for s in range(env.observation_space.n):
        Q_values = [
            sum([prob * (r + gamma * value_table[s_])
                 for prob, s_, r, _ in env.unwrapped.P[s][a]])
            for a in range(env.action_space.n)
        ]
        policy[s] = np.argmax(np.array(Q_values))

    return policy


def value_iteration(env, num_iterations=1000, threshold=1e-20, gamma=1.0):
    optimum_value_function = optimum_value(env, num_iterations, threshold, gamma)
    optimal_policy = extract_policy(optimum_value_function, env, gamma)
    print(optimal_policy)
    return optimal_policy
```

```python
import gymnasium as gym
import time

env = gym.make("FrozenLake-v1", render_mode="human")

policy = value_iteration(env, num_iterations=1000, threshold=1e-20, gamma=1.0)

state, info = env.reset()
env.render()

num_timesteps = 100  # safety cap in case the agent never reaches a terminal state

for t in range(num_timesteps):
    action = int(policy[state])
    state, reward, done, truncated, info = env.step(action)
    print(f'Time Step {t + 1}')
    env.render()

    if done or truncated:
        print(f'Reached Terminal State {state} at time step {t + 1}')
        time.sleep(5)  
        break

env.close()
```

## Policy Iteration

```python
import numpy as np


def compute_value_function(policy, env, num_iterations=1000, threshold=1e-10, gamma=1.0):
    # Initialize value table
    value_table = np.zeros(env.observation_space.n)

    for i in range(num_iterations):
        updated_value_table = np.copy(value_table)

        for s in range(env.observation_space.n):
            a = policy[s]
            value_table[s] = sum([
                prob * (r + gamma * updated_value_table[s_])
                for prob, s_, r, _ in env.unwrapped.P[s][a]
            ])

        if np.sum(np.fabs(updated_value_table - value_table)) <= threshold:
            break

    return value_table


def extract_policy(value_table, env, gamma=1.0):
    policy = np.zeros(env.observation_space.n)

    for s in range(env.observation_space.n):
        Q_values = [
            sum([prob * (r + gamma * value_table[s_])
                 for prob, s_, r, _ in env.unwrapped.P[s][a]])
            for a in range(env.action_space.n)
        ]
        policy[s] = np.argmax(np.array(Q_values))

    return policy


def policy_iteration(env, num_iterations=1000, threshold=1e-10, gamma=1.0):
    # Start with an arbitrary policy (all zeros = always take action 0)
    policy = np.zeros(env.observation_space.n)

    for i in range(num_iterations):
        # Step 1: Policy evaluation - compute value function for the current policy
        value_function = compute_value_function(policy, env, num_iterations, threshold, gamma)

        # Step 2: Policy improvement - extract a new, greedy policy from that value function
        new_policy = extract_policy(value_function, env, gamma)

        # Step 3: Check for convergence - has the policy stopped changing?
        if np.array_equal(new_policy, policy):
            print(f'Policy converged at iteration {i + 1}')
            break

        policy = new_policy

    return policy
```

```python
import gymnasium as gym
import time

env = gym.make("FrozenLake-v1", render_mode="human")

policy = policy_iteration(env, num_iterations=1000, threshold=1e-10, gamma=1.0)
state, info = env.reset()
env.render()

num_timesteps = 100  # safety cap in case the agent never reaches a terminal state

for t in range(num_timesteps):
    action = int(policy[state])
    state, reward, done, truncated, info = env.step(action)
    print(f'Time Step {t + 1}')
    env.render()

    if done or truncated:
        print(f'Reached Terminal State {state} at time step {t + 1}')
        time.sleep(5)  
        break

env.close()
```
